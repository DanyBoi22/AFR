
#### 3. Neustart von Nodes, `rospy` vs. `roscpp` und ROS-Kommunikationsmodell

> **Protokoll-Aufgabe:** Beenden und starten Sie `talker` oder `listener` neu. Starten Sie anschließend `rosrun rospy_tutorials talker` zusammen mit `rosrun roscpp_tutorials listener`. Worin liegt der Unterschied und was sagt dies über die Kommunikation von ROS aus?

**Beobachtungen:**
- **Neustart des `listener`-Nodes:** Während der `listener` beendet ist, gehen alle in diesem Zeitraum vom `talker` gesendeten Nachrichten für diesen Subscriber verloren. Nach dem Neustart werden erst wieder neu veröffentlichte Nachrichten empfangen.
- **Kombination von `rospy_tutorials talker` (Python) und `roscpp_tutorials listener` (C++):** Im Vergleich zum vorherigen Versuch ist der Publisher nun in Python (`rospy`) statt in C++ (`roscpp`) implementiert. Der C++-`listener` empfängt dennoch nahtlos Nachrichten vom Python-`talker`. Werden beide `talker` (`roscpp` und `rospy`) gleichzeitig gestartet, empfängt der `listener` Nachrichten von beiden Sendern.
- **Start zweier `roscpp`-`listener`:** Werden zwei `roscpp_tutorials/listener`-Instanzen mit identischem Node-Namen (`/listener`) gestartet, wird die ältere Instanz automatisch beendet (*shut down*), da Node-Namen in ROS eindeutig sein müssen (sofern kein anonymer Name erzeugt wird).

**Fazit zur ROS-Kommunikation:**  
ROS basiert auf einem **entkoppelten, sprachunabhängigen Publish-Subscribe-Kommunikationsmodell** (Many-to-Many). Mehrere Publisher und Subscriber können über dasselbe Topic (`/chatter`) Daten austauschen, ohne direkt voneinander zu wissen oder in derselben Programmiersprache geschrieben zu sein – entscheidend sind lediglich ein gemeinsamer Topic-Name und ein kompatibler Nachrichtentyp.


#### 4. Aufbau der Python-Skripte `talker.py` und `listener.py`

> **Protokoll-Aufgabe:** Wie sind die beiden Skripte aufgebaut und wie kommunizieren die Nodes miteinander?

**Aufbau und Funktionsweise:**
- **`talker.py` (Publisher):** Registriert ein Publisher-Objekt für das Topic `chatter` mit dem Nachrichtentyp `std_msgs/String` (`queue_size=10`) und initialisiert den Node mit `anonymous=True` (hängt eine eindeutige Kennung an den Namen an). In einer `while`-Schleife wird mit einer Frequenz von **10 Hz** (`rospy.Rate(10)`) ein String inklusive Zeitstempel erzeugt, geloggt und über `pub.publish(hello_str)` veröffentlicht.
- **`listener.py` (Subscriber):** Initialisiert ebenfalls einen eindeutig benannten Node (`anonymous=True`), abonniert über `rospy.Subscriber('chatter', String, callback)` das Topic `chatter` und hält das Skript mittels `rospy.spin()` aktiv. Trifft eine neue Nachricht ein, wird automatisch die `callback(data)`-Funktion aufgerufen und `data.data` ausgegeben.
- **Kommunikation:** Beide Nodes melden sich beim ROS-Master für das Topic `/chatter` an. Anschließend sendet der Publisher Nachrichten asynchron an alle registrierten Subscriber dieses Topics.

**Quellcode-Auszug (`talker.py`):**

```python
def talker():
    pub = rospy.Publisher('chatter', String, queue_size=10)
    rospy.init_node('talker', anonymous=True)
    rate = rospy.Rate(10)  # 10 Hz
    while not rospy.is_shutdown():
        hello_str = "hello world %s" % rospy.get_time()
        rospy.loginfo(hello_str)
        pub.publish(hello_str)
        rate.sleep()
```

**Quellcode-Auszug (`listener.py`):**

```python
def callback(data):
    rospy.loginfo(rospy.get_caller_id() + 'I heard %s', data.data)


def listener():
    # anonymous=True vergibt einen eindeutigen Node-Namen,
    # sodass mehrere Listener gleichzeitig laufen können.
    rospy.init_node('listener', anonymous=True)
    rospy.Subscriber('chatter', String, callback)

    # spin() verhindert das Beenden des Skripts, bis der Node gestoppt wird
    rospy.spin()
```


## 1.2.4. Kinematik

Gemäß Anhang B wurde das ROS-Paket `urdf_example` mit der Modellbeschreibung `example.xacro` (`m2wr` – zweirädriger Roboter) sowie der Launch-Datei `spawn.launch` erstellt und in Gazebo gestartet.

#### 1. Aufbau der URDF-Beschreibung

> **Protokoll-Aufgabe:** Was ist in der URDF-Beschreibung zu sehen? Welche Abschnitte gibt es in URDF-Dateien?

Die Datei beschreibt im XML-Format ein zweirädriges mobiles Robotermodell (`<robot name="m2wr">`) bestehend aus einem Chassis mit Stützrad (`link_chassis`) sowie einem linken und rechten Antriebsrad (`link_left_wheel`, `link_right_wheel`). Eine URDF-Datei gliedert sich in folgende Hauptabschnitte:

- **`<robot>`:** Wurzelelement des gesamten Robotermodells.
- **`<material>`:** Definition wiederverwendbarer Materialien und RGBA-Farbwerte (z. B. `black`, `blue`, `orange`, `white`).
- **`<link>`:** Beschreibung der einzelnen starren Körper (Bauteile) mit visueller Geometrie (`<visual>`), Kollisionsgeometrie (`<collision>`) und Trägheitseigenschaften (`<inertial>`).
- **`<joint>`:** Beschreibung der mechanischen Verbindungen zwischen jeweils zwei Links (`parent` und `child`) inklusive Gelenktyp, Position/Orientierung und Bewegungsachse.
- **`<gazebo>`:** Gazebo-spezifische Erweiterungen (Materialzuweisungen für das Gazebo-Rendering sowie Simulations-Plugins wie der Differentialantriebs-Controller).

#### 2. Elemente zur Beschreibung der Kinematik

> **Protokoll-Aufgabe:** Welche Elemente dienen zur Beschreibung der Kinematik des Roboters?

Die Kinematik beschreibt die geometrische Struktur und die möglichen Bewegungen des Roboters (ohne Kräfte). Dafür werden folgende URDF-Elemente verwendet:

- **`<link>`:** Repräsentiert die starren Glieder und deren lokale Koordinatensysteme.
- **`<joint type="...">`:** Definiert die Verbindung zwischen `<parent link="..."/>` und `<child link="..."/>` sowie den Freiheitsgrad (z. B. `continuous` für endlose Raddrehung, `revolute`, `prismatic`, `fixed`).
- **`<origin xyz="..." rpy="..."/>`:** Legt die räumliche Transformation (Translation und Orientierung als Roll, Pitch, Yaw) zwischen den Koordinatensystemen fest.
- **`<axis xyz="..."/>`:** Bestimmt die Dreh- oder Translationsachse des Gelenks (bei den Rädern die Y-Achse).
- **`<limit>`:** Optional zur Begrenzung von Verfahrwegen, Drehwinkeln, Geschwindigkeiten oder Drehmomenten.

#### 3. Zusätzliche Angaben für die Simulation

> **Protokoll-Aufgabe:** Was sind Angaben, die zusätzlich für die Simulation gebraucht werden?

Für eine physikalische und grafische Simulation in Gazebo reichen rein kinematische Angaben nicht aus. Zusätzlich erforderlich sind:

- **Trägheits- und Masseneigenschaften (`<inertial>`):** Masse (`<mass>`), Schwerpunktlage (`<origin>`) und Trägheitstensor (`<inertia ixx="..." .../>`) für die Dynamikberechnung.
- **Kollisionsgeometrie (`<collision>`):** Vereinfachte geometrische Hüllkörper (`<box>`, `<cylinder>`, `<sphere>`) zur Kollisionserkennung und Bodenkontaktberechnung.
- **Visuelle Darstellung (`<visual>` und `<material>`):** Geometrie und Farben für das Rendering.
- **Gazebo-spezifische Eigenschaften (`<gazebo>`):** Simulationsmaterialien (z. B. `<material>Gazebo/Black</material>`), Reibungskoeffizienten (`mu1`, `mu2`) sowie **Plugins** (z. B. `libgazebo_ros_diff_drive.so` zur Anbindung von `/cmd_vel` und `/odom` an die simulierten Radgelenke).
- **Gelenkdynamik (`<dynamics>`):** Dämpfung (`damping`) und Reibung (`friction`) der Gelenke.
- **Launch-Konfiguration (`spawn.launch`):** Startet die Gazebo-Welt, lädt die Xacro-Datei in den Parameter `robot_description` und erzeugt das Modell über `gazebo_ros/spawn_model` an einer Startposition `(x, y, z)`.

**Verwendete `spawn.launch`:**

```xml
<?xml version="1.0" encoding="UTF-8"?>
<launch>
    <arg name="x" default="0.0"/>
    <arg name="y" default="0.0"/>
    <arg name="z" default="0.5"/>

    <include file="$(find gazebo_ros)/launch/empty_world.launch">
        <arg name="world_name" value="$(find turtlebot3_gazebo)/worlds/empty.world"/>
        <arg name="use_sim_time" value="true"/>
        <arg name="gui" value="true"/>
        <arg name="headless" value="false"/>
        <arg name="debug" value="false"/>
    </include>

    <param name="robot_description"
           command="$(find xacro)/xacro --inorder $(find urdf_example)/urdf/example.xacro"/>

    <node pkg="gazebo_ros" type="spawn_model" name="spawn_urdf"
          args="-urdf -model m2wr -x $(arg x) -y $(arg y) -z $(arg z) -param robot_description"/>
</launch>
```


#### 4. Anpassung der URDF-Datei und Auswirkung in Gazebo

> **Protokoll-Aufgabe:** Verändern Sie Einträge in der URDF-Datei und prüfen Sie, was sich im Verhalten und/oder der Darstellung in Gazebo dadurch geändert hat! Verändern Sie die Farbe des Gazebo-Elementes `link_chassis` zu schwarz!

**Durchgeführte Änderungen in `example.xacro`:**
1. **Farbe von `link_chassis`:** Über das zugehörige Gazebo-Tag (`<gazebo reference="link_chassis"><material>Gazebo/Black</material></gazebo>`) wurde die Farbe des Chassis in Gazebo auf **Schwarz** geändert (diese Angabe überschreibt in Gazebo das `<material name="blue"/>` aus dem `<visual>`-Block).
2. **Größe des Chassis (`link_chassis`):** Die Länge der Box in `<collision>` und `<visual>` wurde von `0.5 m` auf **`5.0 m`** vergrößert (`<box size="5.0 0.3 0.07"/>`).
3. **Radradius (`link_right_wheel` und `link_left_wheel`):** Der Zylinderradius beider Räder wurde in `<collision>` und `<visual>` stark auf **`1.0 m`** vergrößert (`<cylinder length="0.04" radius="1.0"/>`).

**Beobachtete Auswirkungen in Gazebo:**  
Das Chassis wird wie gewünscht **schwarz** dargestellt.
